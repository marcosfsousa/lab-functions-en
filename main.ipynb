## 2. Now write a function that returns the largest element on a list

```python
def greatest(lst):
    # initialize a variable with the first element of the list
    max_value = lst[0]
    for element in lst:
        # we start a comparison between the element at the first index and compare to the next element
        if max_value < element:
            # store the biggest element encountered so far in the loop
            max_value = element
    return max_value
```

```python
# This will test your function 
test_greatest(greatest)
```

## 6. Write a function that returns the factorial of a number.

```python
def factorial(n):
    result = 1
    for element in range(1, n+1):
        result *= element
    return result

factorial(6)
```

```python
#factorial formula
#n! = n * ( n - 1 ) *...*1

# This code defines a function called "factorial" which takes an input "n". The function uses a for loop to iterate through the range of numbers 
# from 1 to n+1. For each number in that range, it multiplies the current value of x by the number in the range. At the end of the loop, 
# the function returns the final value of x, which will be the factorial of the input number "n".

# The Factorial of a positive integer n is the product of all positive integers less than or equal to n. 
# For example, the factorial of 6 (written "6!") is 6 * 5 * 4 * 3 * 2 * 1 = 720.

# So this function takes an input of any positive integer, and returns the factorial of that number.
```

## 7. Write a function that takes a list and returns a list of the unique values.

`NOTE: You cannot use set. 🤔`

```python
def unique(lst_un):
    result_list = []
    for element in lst_un:
        if element not in result_list:
            result_list.append(element)
    return result_list
```

## 8. Write a function that returns the mode of a list, i.e.: the element that appears the most times.
`NOTE: You should not use count... 🧐`

```python
def mode_counter(arr):
    # Initializing the variables we need to complete the exercise
    unique_dict = {}
    # This variable is initialized as None given that we don't know if the list will have mixed mode types.
    most_repeated_key = None
    max_count = 0

    # We start by iterating our list
    for element in arr:
        # Membership test to our dictionary and start populating it
        if element in unique_dict:
            # if the key is present we add 1 to its count
            unique_dict[element] += 1
            # if the key is not present we create it and initialize its value to 1
        elif element not in unique_dict:
            unique_dict.setdefault(element, 1)
    # This loop will verify what is the highest frequency value in our new dictionary
    for key, values in unique_dict.items():  
        # we test if the value found is bigger than our last known max value
        if values > max_count:
            # if true we store it
            max_count = values
            # and store its respective key
            most_repeated_key = key
    return most_repeated_key

# This code will return the first instance of the highest number found and not deal with duplicate highest numbers. That validation wasn't added as we felt it was not part of the exercise.
```

## 9. Write a function that calculates the standard deviation of a list.
`NOTE: Do not use any libraries or already built functions. 😉`

```python
def st_dev(list_sd):
    mean = 0
    running_total = 0
    sqrd_deviation = 0
    std_deviation = 0

    # Getting the mean by adding all values inside a list and divide it by length of list
    for element in list_sd:
        running_total = running_total + element    
    mean = running_total / len(list_sd)
    
    # Getting the sum of squared deviations from the mean (negatives don't cancel positives and larger deviations have extra weight)
    for element in list_sd:
        sqrd_deviation += ((element - mean)**2)
    
    # Getting the standard deviation by dividing the sum of squared deviations by (list length - 1) and taking square root.
    std_deviation = (sqrd_deviation / (len(list_sd) - 1)) ** 0.5    
    return std_deviation

# Very difficult exercise given the mathematical formulas required to complete it. 
```

## 10. Write a function to check if a string is a pangram, i.e.: if it contains all the letters of the alphabet at least once. Mind that the strings may contain characters that are not letters.

```python
import string
string.ascii_lowercase
alphabet = list(string.ascii_lowercase)

def pangram(string):
    string = string.lower()
    for element in alphabet:
        if element not in string:
            return False
    return True
```

## 11. Write a function that receives a string of comma separated words and returns a string of comma separated words sorted alphabetically.

`NOTE: You may use sorted but not split and definitely no join! 🤪`

```python
def sort_alpha(string):
    # Initialize variables as we need them to hold temporary values
    word_list = []
    word_string = ''
    clean_string = ''
    current_word = ''
    # Iterate over elements in string argument
    for element in string:
        # Test for the known delimiter - comma
        if element != ',':
            # Add characters to form a word until delimiter is found
            current_word += element
        else:
            # Append words to our list of words
            word_list.append(current_word)
            # Reset word holder to be able to fill a new word in the next iteration of the loop
            current_word = ''
    # Append the final word, which is not followed by a delimiter and is not included in the loop
    word_list.append(current_word)
    # Sorted the word list alphabetically
    word_list = sorted(word_list)
    # Rebuild the sorted list into a string
    for element in word_list:
        # Add a comma after each word (element)
        word_string += element + ','
    # Clean the string of a trailing delimiter by removing the last character (since we know it's there)
    clean_string = word_string[:-1]
    return clean_string
```

```python
# This will test your function 
test_alpha(sort_alpha)
```

## 12. Write a function to check if a given password is strong (at least 8 characters, at least one lower case, at least one upper case, at least one number and at least one special character). It should output True if strong and False if not.

```python

def check_pass(password):
    # Initialize flag variables
    has_lowercase = False
    has_uppercase = False
    has_digit = False
    has_special_char = False
    
    # First test condition that can be verified out of the loop since the length never changes allowing us to fail fast
    if len(password) < 8:
        return False
    
    # iterating over chars in string
    for element in password:
        # test if it has lowercase chars
        if element.islower():
            has_lowercase = True
        # test if it has uppercase chars
        if element.isupper():
            has_uppercase = True
        # test if it has digits
        if element.isdigit():
            has_digit = True
        # test if it has special characters a.k.a. non-alphanumeric 
        if not element.isalnum(): 
            has_special_char = True 
    # if any of the four flags are false the password doesn't meet the criteria   
    if not (has_lowercase and has_uppercase and has_digit and has_special_char):
        return False
    else:
        return True
```

```python
# This will test your function 
test_pass(check_pass)
```
